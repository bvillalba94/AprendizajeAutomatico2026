```python
import pandas as pd

# Se cargan los datos de ventas desde el CSV

ventas = pd.read_csv("ventas.csv")

print("Primeras filas del archivo de ventas:")
print(ventas.head())

# Se cargan los datos de clientes desde el JSON

clientes = pd.read_json("clientes.json")

# Se muestra la cantidad total de clientes
print("\nCantidad total de clientes:")
print(len(clientes))

# Se cargan los datos de inventario desde Excel

inventario = pd.read_excel("inventario.xlsx")

# Se calcula el promedio
promedio_stock = inventario["stock"].mean()

print("\nPromedio de stock disponible:")

print(promedio_stock)
```
