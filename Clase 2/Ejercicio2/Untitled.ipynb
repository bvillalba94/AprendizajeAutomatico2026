```python
import pandas as pd

import matplotlib.pyplot as plt

# Cargar dataset

df = pd.read_csv('Automobile.csv')

# Mostrar primeras filas

print(df.head())

# Información general
print(df.info())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['mpg'], bins=15, color='skyblue', edgecolor='black')
plt.title('Distribución del Consumo de Combustible (MPG)')
plt.xlabel('MPG')

plt.ylabel('Frecuencia')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['weight'], df['mpg'], color='steelblue')

plt.title('Relación entre Peso y Consumo de Combustible')
plt.xlabel('Peso del automóvil')
plt.ylabel('MPG')
plt.grid(True)

plt.show()
```

```python
df['origin_num'] = df['origin'].map({
    'usa': 1,
    'europe': 2,
    'japan': 3
})

plt.figure(figsize=(8,5))

plt.boxplot(
    [df['cylinders'], df['origin_num']],
    tick_labels=['Cylinders', 'Origin']
)

plt.title('Boxplot de Cylinders y Origin')
plt.ylabel('Valores')
plt.show()
```

```python
distribucion = pd.crosstab(df['model_year'], df['origin'])

distribucion.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Distribución de Automóviles por Año y Origen')
plt.xlabel('Año del Modelo')
plt.ylabel('Cantidad de Automóviles')
plt.legend(title='Origen')
plt.tight_layout()

plt.show()
```
