```python
import pandas as pd

import matplotlib.pyplot as plt

# Cargar dataset

df = pd.read_csv('Automobile.csv')

# Mostrar primeras filas

print(df.head())

# Información general
print(df.info())
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['weight'], df['mpg'], color='steelblue')

plt.title('Relación entre Peso y Consumo de Combustible')
plt.xlabel('Peso del automóvil')
plt.ylabel('MPG')
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['mpg'], bins=15, color='skyblue', edgecolor='black')
plt.title('Distribución del Consumo de Combustible (MPG)')
plt.xlabel('MPG')

plt.ylabel('Frecuencia')

plt.show()
```

```python
df[['cylinders', 'origin']].boxplot(figsize=(8,5))

plt.title('Boxplot de Cylinders y Origin')
plt.ylabel('Valores')
plt.show()
```
